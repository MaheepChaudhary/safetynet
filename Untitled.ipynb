```python
import pickle as pkl
from transformers import AutoTokenizer

# Load the data
with open("safetynet/safetynet/spylab.pkl", "rb") as f:
    data = pkl.load(f)

# Check for empty completions
print("=== Data Structure ===")
print(f"Keys: {data.keys()}")
print(f"Total samples: {len(data['prompt'])}")

# Count by label
from collections import Counter
label_counts = Counter(data['label'])
print(f"Label distribution: {label_counts}")

# Check for empty completions
empty_completions = sum(1 for c in data['completion'] if not c or len(c.strip()) == 0)
short_completions = sum(1 for c in data['completion'] if len(c.strip()) < 10)
print(f"\nEmpty completions: {empty_completions}")
print(f"Short completions (<10 chars): {short_completions}")

# Check by label
print("\n=== By Label ===")
for label in set(data['label']):
    indices = [i for i, l in enumerate(data['label']) if l == label]
    completions = [data['completion'][i] for i in indices]
    empty = sum(1 for c in completions if not c or len(c.strip()) == 0)
    short = sum(1 for c in completions if len(c.strip()) < 10)
    print(f"{label}: {len(indices)} samples, {empty} empty, {short} short (<10 chars)")

# Show sample of harmful data
print("\n=== Sample Harmful Data ===")
harmful_indices = [i for i, l in enumerate(data['label']) if l == 'harmful'][:3]
for i in harmful_indices:
    print(f"\n--- Sample {i} ---")
    print(f"Prompt: {data['prompt'][i][:100]}...")
    print(f"Completion: '{data['completion'][i][:100]}...' (len={len(data['completion'][i])})")

# Test tokenization for one sample
print("\n=== Tokenization Test (llama2) ===")
tokenizer = AutoTokenizer.from_pretrained("meta-llama/Llama-2-7b-chat-hf")
tokenizer.pad_token = tokenizer.eos_token

chat_template = "<s>[INST] {prompt} [/INST] {completion} </s>"
prompt_template = "<s>[INST] {prompt} [/INST]"

for i in harmful_indices[:2]:
    prompt = data['prompt'][i]
    completion = data['completion'][i]
    
    full_text = chat_template.format(prompt=prompt, completion=completion)
    prompt_text = prompt_template.format(prompt=prompt)
    
    full_enc = tokenizer(full_text)
    prompt_enc = tokenizer(prompt_text)
    
    completion_tokens = len(full_enc["input_ids"]) - len(prompt_enc["input_ids"])
    print(f"\nSample {i}:")
    print(f"  Full tokens: {len(full_enc['input_ids'])}")
    print(f"  Prompt tokens: {len(prompt_enc['input_ids'])}")
    print(f"  Completion tokens: {completion_tokens}")
    if completion_tokens <= 0:
        print(f"  ⚠️  WARNING: No completion tokens!")
```

```python
import torch

# Use the correct path with full model name
dataset_path_prefix = "safetynet/safetynet/spylab/Llama-2-7B-chat-hf"

print("=== Loading cached datasets ===")
normal_dataset = torch.load(f"{dataset_path_prefix}_normal_dataset.pt")
backdoor_dataset = torch.load(f"{dataset_path_prefix}_backdoor_dataset.pt")

print(f"Normal dataset size: {len(normal_dataset)}")
print(f"Backdoor dataset size: {len(backdoor_dataset)}")

print("\n=== Checking Normal Dataset Labels ===")
for i in range(min(5, len(normal_dataset))):
    labels = normal_dataset[i]['labels']
    if not isinstance(labels, torch.Tensor):
        labels = torch.tensor(labels)
    valid_labels = labels[labels != -100]
    print(f"Normal {i}: total={len(labels)}, valid={len(valid_labels)}, tokens={valid_labels[:5].tolist() if len(valid_labels) > 0 else 'NONE'}")

print("\n=== Checking Backdoor Dataset Labels ===")
for i in range(min(5, len(backdoor_dataset))):
    labels = backdoor_dataset[i]['labels']
    if not isinstance(labels, torch.Tensor):
        labels = torch.tensor(labels)
    valid_labels = labels[labels != -100]
    print(f"Backdoor {i}: total={len(labels)}, valid={len(valid_labels)}, tokens={valid_labels[:5].tolist() if len(valid_labels) > 0 else 'NONE'}")

# Count samples with ALL -100 labels
print("\n=== Counting samples with ALL -100 labels ===")
backdoor_all_masked = sum(1 for s in backdoor_dataset if (torch.tensor(s['labels']) == -100).all())
normal_all_masked = sum(1 for s in normal_dataset if (torch.tensor(s['labels']) == -100).all())
print(f"Normal: {normal_all_masked}/{len(normal_dataset)} have ALL -100")
print(f"Backdoor: {backdoor_all_masked}/{len(backdoor_dataset)} have ALL -100")
```
